# Q7. Sum of weights

```python
cars_from_asia = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
cars_from_asia
```

```python
X = cars_from_asia.values
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```

```python
result = w.sum()
print(result)
```
